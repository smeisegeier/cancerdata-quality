```python
import shutil
from pathlib import Path


def _create_section_12_report():
    DIR_REPO = Path("../../Bericht_Erfahrungen_mit_der_bundesweiten_Erfassung_von_Krebsregisterdaten")
    DIR_SRC = Path("./section-12-report")
    KEEP = {"lizenz", "license", "licence"}

    for item in DIR_REPO.iterdir():
        if item.name == ".git" or item.name.lower() in KEEP:
            continue
        if item.is_dir():
            shutil.rmtree(item)
        else:
            item.unlink()

    shutil.copytree(
        DIR_SRC,
        DIR_REPO,
        ignore=shutil.ignore_patterns("*.py", "__pycache__", ".DS_Store"),
        dirs_exist_ok=True,
    )
```
